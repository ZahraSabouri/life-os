```python
# 2. Write code which can convert any number of days, hours, minutes and seconds into seconds.
#
# Can you do this as a function? Optionally, can you also write code to convert any number of seconds into days,
#  hours, minutes and seconds?
```

```python
# dhms_to_s: takes a number of days, hours, minutes and seconds,
#  and returns the total number of seconds this represents
def dhms_to_s( days, hours, minutes, seconds ):
    just_seconds = ( days * 24 * 60 * 60 ) + ( hours * 60 * 60 ) + ( minutes * 60 ) + seconds
    #                Each day is 24 hours; Each hour is 60 minutes; Each minute is 60 seconds
    return just_seconds

# Let's try our function:
#  Choose a number of days, hours, minutes and seconds (we could pass values straight
#   into our function, but this is easier to alter and run many times)
days = 1
hrs = 2
mins = 30
secs = 15
#  Now, we can call our function, pass in the d,h,m,s variables, and store the result as "ans"
ans = dhms_to_s( days, hrs, mins, secs )
#  Let's print out the result, in a way which is easy to read (and check)
print( days, "d ", hrs, "h ", mins, "m ", secs, "s = ", ans, "seconds" )
```

```python
# Optional part:

# s_to_dhms: takes a number of seconds, and returns the total number
#  of days, hours, minutes and seconds this represents
def s_to_dhms( seconds ):
    days = seconds // ( 24 * 60 * 60 )    # Take the integer division of total seconds against 24d x 60m x 60s
    seconds -= days * 24 * 60 * 60        # Subtract this from the seconds we currently have (we've accounted for it)
    hrs  = seconds // ( 60 * 60 )         # Take the integer division of current seconds against 60m x 60s
    seconds -= hrs * 60 * 60              # Subtract this from the current total seconds (it's accounted for)
    mins = seconds // 60                  # Take the integer division of what remains against 60s (to get minutes)
    seconds -= mins * 60                  # Take that many minutes away from the current seconds count (we have it)
    return days, hrs, mins, seconds       # Now we have our days, hours, minutes and seconds, return them

# Let's try our function:
#  Choose a total number of seconds to convert into days, hours, minutes and seconds
secs = 95415
#  Call our function, passing that number of seconds in, and storing the 4-part tuple as d, h, m and s
d, h, m, s = s_to_dhms( secs )
#  Let's print out the result in a way which is easy to read (and check)
print( secs, "seconds = ", d, "d ", h, "h ", m, "m ", s, "s " )
```
