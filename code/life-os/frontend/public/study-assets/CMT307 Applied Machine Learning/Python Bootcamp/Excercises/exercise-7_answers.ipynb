```python
# 1. Write code which can create a 2D NumPy array of a given size, which is filled with zeros and has ones all
#  around its border (i.e. left and right columns, and top and bottom rows). Is there a very compact way to achieve
#  this? 
```

```python
import numpy as np         # We must import NumPy before we can use it; it's convenient to use the "np" abbreviation

size_m = 5                             # Let's make a variable to store our number of rows (matrix "m" dimension)
size_n = 5                             # Let's make a variable to store our number of columns (matrix "n" dimension)

arr = np.ones( ( size_m, size_n ) )    # Create a new m-by-n array, and initialise it with all ones
arr[ 1 : -1 , 1 : -1 ] = 0             # Starting at the second element, and finishing at the second-but-last
                                       #  element, fill all rows _and_ columns in-between with zeros

print( arr )                           # Print out the matrix, so we can see the result
```

```python
# 2. Noughts and Crosses: create a Python program which:
#  (a) Creates an empty 3x3 NumPy array of a suitable datatype;
#  (b) Accepts input from a player, allowing them to specify a row and column in the matrix (array);
#  (c) Checks to see whether there is already a nought or a cross in that cell, and, if not, adds one;
#  (d) Checks to see whether the game has been won (that is, there are either three crosses or three noughts in any
#   diagonal, horizontal or vertical row);
#  (e) If the game has not been won, repeats from part (b) so that the player can enter another cell.
#
# Optionally, you could choose to print out the NumPy matrix after each turn (so that the players can see the state
#  of the playing board), and to keep track of whose turn it is (printing out "Noughts" or "Crosses" each time
#  input is being requested, depending on which token is being played).
```

```python
# 2(a)
grid = np.zeros( ( 3, 3 ), dtype = np.uint8 ) # We can't use a bool (spaces can be empty/o/x), but an unsigned
                                              #  8-bit integer can work (and it won't be overly wasteful)
EMPTY  = 0                                    # Let's make three variables which we can use to track the contents
NOUGHT = 1                                    #  of our game grid; a zero means "empty", a one means "a nought", and
CROSS  = 2                                    #  a two means "a cross"

# 3(e): We'll need this to keep track of whether the game has ended, or still needs to run
won = 0                                       # Let's use a variable to track win status (and which token won)
# 3(c): We need to initialise this before we start, or we'll only ever see one token in our grid!
current = NOUGHT                              # We'll need some way to keep track of which token we're entering!
                                              #  Let's start with a "nought" for now (and swap for "cross" each turn)
# 3(e): Repeat the process if nobody has won
while not won:                                # Add a "while" loop to control repeating (we can do this because
                                              #  not 0 = True, not 1 = False, and not 2 = False)
    
    # "Optional" part: we can print out which token is in play currently:
    if current == NOUGHT: print( "[Noughts' turn]" )
    else:                 print( "[Cross' turn]" )
    # 2(b)
    pos = input( "Enter position to fill (as row,column): " )  # Ask the user to enter e.g. "2,1" for row=2, col=1
    pos = pos.split( "," )                        # Split the user's input around the comma (note: we _should_ check
                                                  #  to make sure the input is correctly formed; for now, this will do)
    row = int( pos[ 0 ] ) - 1                     # Assuming the input was correct, element 0 will now be our "row"
    col = int( pos[ 1 ] ) - 1                     #  number, and element 1 will be our "column" number. We're
                                                  #  subtracting 1 each time so we can enter "1" for index "0", etc.
    # 3(c)
    if grid[ row, col ] == EMPTY:                 # Check to see whether this cell is empty;
        grid[ row, col ] = current                #  if so, set it to contain the current token,
        #current = ( NOUGHT - ( current - NOUGHT ) ) + NOUGHT  # We can do this to flip the state; it's not very readable
        if current == NOUGHT: current = CROSS     # This is more intuitive, though; since we've just used the current
        else:                 current = NOUGHT    #  token, change it to the next one (ready for the next play)
    else:                                                                  # If the square is full,
        if grid[ row, col ] == NOUGHT: contains = "nought"                 #  Let's make a string saying what it contains
        else:                          contains = "cross"                  #  ...
        print( "The square already contains a %s; try again." % contains ) #  ...and tell the user

        
    # "Optional" part:
    #print( grid )                             # Show the player the grid... but we could make this more meaningful,
    #                                          #  i.e. have "x" or "o" characters instead of "1" or "2"
    # "Optional" part: Okay, let's make it more meaningful (but, naturally, our code becomes more complex!):
    for m in range( -1, 3 ):                                        # Iterate over -1 (header), then 0-2 (our rows)
        if m == -1:                                                 #  If we're on the first row...
            print( " Cols: ",     end="" )                          #   ...print "cols" to show column numbers follow
        else:                                                       #  Otherwise, print the row number:
            print( "Row %d: " % ( m + 1 ), end="" )                 #   (note we use 'end=""' to prevent the newline)
        for n in range( 3 ):                                        #  Iterate over 0-2 (our columns)
            if m == -1:                                             #  If we're on the first row...
                print( "%d " % ( n + 1 ), end="" )                  #   ...print the column number instead
            else:                                                   #  Otherwise:
                if   grid[ m, n ] == NOUGHT: print( "o ", end="" )  #   If a "nought" is present, print "o"
                elif grid[ m, n ] == CROSS:  print( "x ", end="" )  #   Else, if a "cross" is present, print "x"
                else:                        print( "- ", end="" )  #   Else, print a dash (to mark an empty space)
        print()                                                     #  Print a newline (so we can start the next row)


    # 3(d)
    # We know we broadly have four conditions to check, each of which will signify a win:
    #  (1) A column containing all the same tokens (there are three of those to check);
    #  (2) A row containing all the same tokens (there are three of those to check, too);
    #  (3) A single diagonal from top-left to base-right containing the same tokens; and
    #  (4) A single diagonal from top-right to base-left containing the same tokens.
    # We can check for (1) and (2) using a for loop (one iteration for each row, or column),
    #  and we can check the diagonals directly (there are only two we need to consider).
    # If ANY of these conditions are true, the game has been won;
    #  to make things easier, let's make a boolean variable called "win", and use the
    #  boolean OR operation at each test (so that ANY of the conditions above evaluating to
    #  True will set our "win" flag to True, even if later conditions evaluate to False):
    for token in [ NOUGHT, CROSS ]:                                     # Let's check for NOUGHT first, then CROSS
        win = False                                                     #  Here's our "win" flag (False at first)
        # Check each row and column (conditions 1 & 2 above):
        for index in range( 3 ):                                        #  Now, let's index over each row and column
            win |= np.sum( grid[ index, : ] == token ) == 3             # These treat each "True" as a "1"; if we get
            win |= np.sum( grid[ :, index ] == token ) == 3             #  3 Trues for a row (or a column), it means
        # Check both diagonals (conditions 3 and 4 above):              #  that token is in ALL elements (i.e. a win)
        win |= np.sum( grid[ range(3), range(3) ] == token ) == 3       # Note this code (left) fetches each diagonal,
        win |= np.sum( grid[ range(3), range(2,-1,-1) ] == token ) == 3 #  using lists [0,1,2],[0,1,2] for the upper,
                                                                        #  and [0,1,2],[2,1,0] for the lower line.                                                                        
        if win:                                                         # Were any of these True?
            won = token                                                 #  If so, set "win" to the token which won

# 3(e): Why not also tell the user when the win has happened? It's not part of the question, but it will be useful
if won == NOUGHT:
    print( "Noughts won!" )
elif won == CROSS:
    print( "Crosses won!" )
```

```python
# 3. Write code which creates a 3x3x3 array where the first element is 0, the last is 6.5, and the elements in
#  between are all linearly-spaced, increasing values.
#  (a) Which index of your array contains the value "5"?
#  (b) Using your array, and any NumPy methods you believe are useful, identify which index of a _1D version of
#   that array_ would contain the value "5".
#  (c) Write standard Python looping statements to iterate through the array, and verify that the index you have
#   from (a) actually refers to the correct element.
#  (d) Optionally, adapt your code (or write new code) which will allow you to perform the above operations on
#   _any_ dimension of 3D array, and for any sought value (have your code gracefully handle the situation where
#   that value is not present in the array).
```

```python
arr = np.linspace( 0, 6.5, 3*3*3 ).reshape( 3, 3, 3 )    # linspace() allows us to start at 0, end at 6.5, and
                                                         #  spread our values over 3*3*3 = 27 elements in total

# 3(a): I can see that the third square (dimension 1), the first row (dimension 2),
#        and the third column (dimension 3) contain the value 5.
print( arr )
#       So, the index must be [2,0,2] (-1 for each index, because we start at index 0)
print( arr[ 2, 0, 2 ] )                                  # This is correct; we can see we get the value "5"

# 3(b)
#  We can do this in individual steps...:
arr_1d = arr.reshape( 3*3*3 )                            # Reshape our 3x3x3 array into a 1D array of length 27
mask   = arr_1d == 5                                     # Create a mask, all False except where the value is 5
index  = np.argmax( mask )                               # Use np.argmax() to get the index of the greatest value,
print( arr_1d, mask, index, sep = "\n" )                 #  which happens to be "True" (1, a "False" is equiv. to 0)
#  ...or, we can write a single, compound statement (this seems the better
#      option in this case, as it remains quite readable):
print( np.argmax( arr.reshape( 3*3*3 ) == 5 ) )          # We can reshape, then create a mask (==), then use argmax
                                                         #  on it, all in one expression!
# 3(c)
#  We can use the "arr_1d" reshaped array we created above,   <- but ONLY if we're not using the compound expression!
#   and a for loop (which we can "break" from when the right
#   element is found):
for i in range( arr_1d.shape[ 0 ] ):                     # Iterate over indices 0-26, i.e. dimension 1 (index 0)
    if arr_1d[ i ] == 5:                                 #  Does the element at index i contain the value 5?
        print( "Found 5, at index", i )                  #   If it does, print out the index we found it at...
        break                                            #   ...and then abort our search (we've finished!)
```

```python
# 3(d): We'll adapt our code to work with array dimensions l,m,n,
#        and seek a value v.
#       We should probably also allow a different end value to be
#        chosen (not just 6.5!), but the question didn't ask for that;
#        we can always add it as an extra, if we wish (we won't, for now).

l = 5 #3
m = 5 #3
n = 3 #3
#
v = 3.25

arr = np.linspace( 0, 6.5, l*m*n ).reshape( l, m, n )    # linspace() allows us to start at 0, end at 6.5, and
                                                         #  spread our values over l*m*n elements in total
print( arr )

# 3(b): Previously, we _knew_ the value 5 would be present;
#        now, we're not sure, so we should check that first
arr_1d = arr.reshape( l*m*n )                            # Let's reshape our array; we can use it several times below
found = np.any( arr_1d == v )                            # Let's use np.any() to detect if ANY element contains "v"
if found:                                                # Only if it does will we proceed:
    print( np.argmax( arr_1d == v ) )                    #  Let's create a mask (==), then use np.argmax, all in one
                                                         #   expression
    # 3(c)
    #  We can use the "arr_1d" reshaped array we created above,
    #   and a for loop (which we can "break" from when the right
    #   element is found:
    for i in range( arr_1d.shape[ 0 ] ):                     # Iterate over all our 1D indices (still dimension 1)
        if arr_1d[ i ] == v:                                 #  Does the element at index i contain the value "v"?
            print( "Found %f, at index %d" % ( v, i ) )      #   If it does, print out the index we found it at...
            break                                            #   ...and then abort our search (we've finished!)
# We should also handle the case where found == False, gracefully:
else:
    print( "The value %f was not found in the array." % v )
```
