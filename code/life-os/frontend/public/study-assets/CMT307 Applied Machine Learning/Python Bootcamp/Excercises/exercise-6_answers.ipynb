```python
# EXERCISE 6
#
# An Anagram Finder
#
# An anagram is a word formed by rearranging the letters of another word. We can say one word is an anagram of
#  another if both have the same number of letters (for instance, "cat" and "tar", which both have one "c", one
#  "a", and one "t"). Let's make a program which can find _all_ the anagrams in a given file list of words.
#
```

```python
# 1. First of all, download the list of words "words.txt" from LC. Notice that each word in the file is on a new
#  line. Write code which will load "words.txt" so that each word is an element of a list, and any newline character at the end of the word is removed. Add code which will search this list for the word "data", and print out the index of this word.
#
# NOTE: In order to load the file in Colab, you'll need to add it to your Files list (from the left-hand navigation
#  pane). Go to your list of files, and choose the "Upload" option. Once uploaded, you'll be able to work with your
#  file from within Python as normal. Colab will warn you that the file storage is temporary (you'll have to
#  re-upload it again if you return to your notebook), but that's fine for now (permanent storage should be on a
#  Google Drive, for example).
```

```python
f = open( "words.txt", "r" )               # Open the file "words.txt" in "read" mode (mode string "r")
data = f.read().split()                    # Read ALL the data in, and then use ".split()" to separate it by lines
f.close()                                  # We've finished with the file; we should close it

words = [ word.strip() for word in data ]  # Using a list comprehension to create a new list; we're iterating over
                                           #  each word, and using ".strip()" to get rid of the "\n" each ends in

## TEST CODE: Once I can see the words loaded correctly, I can comment this out
#print( words )
```

```python
# 2. Write code which will iterate over all the words in your list, and use a Counter (or a dictionary, if you
#  prefer) for each to build a histogram of the number of letters in that word (print the word and the resulting
#  counter out each time, so that you can see the results). Now, add a new dictionary, and store your counter for
#  each word in that dictionary (with the word as the key, and the Counter as the value).
```

```python
from collections import Counter          # We must import Counter (from "collections"); it's not available built-in

## First part (we can comment this out in a moment, when we adapt it):
#for word in words:                       # Iterate over all the words, to get one word at a time
#    print( "Word is:", word, "; its letter count is:", Counter( word ) )  #  Print the word, and its Counter(word)
    
# Second part (adapted from the code above; we're just adding dictionary storage):
letters = {}                             # Let's call our dictionary "letters" (really, it's letter_counts)
for word in words:                       # Iterate over all the words, to get one word at a time
    letters[ word ] = Counter( word )    #  Build a Counter for each word, and store it in the dictionary (key=word)
    print( "Word is:", word, "; its letter count is:", letters[ word ] )   #  Print the word, and its Counter(word)
```

```python
# 3. Next, write code which takes your dictionary of words : letter_counts, and creates a new dictionary, this time
#  with a single counter per key, and a list of words which match that counter as its value.
```

```python
anagrams = {}                            # Let's call our dictionary "anagrams"; two words with same count = anagram!
for key in letters.keys():               # Iterate over the keys of our "letters" dictionary (each key is a _word_)
    counter = letters[ key ]             #  Fetch the counter for each key (word) from that dictionary 
    #counter = str( counter )            #  Why don't we just do this? It's to do with _ordering_ (see below...)
    counter = str( sorted( counter.items() ) )  # Here, we should take a few steps (and can compound them):
     # (1) We can get the ".items()" of our counter (e.g. dict_items([('t', 1), ('a', 1)]) for the word "at"), then
     # (2) We can _sort_ them; this happens to give us a list (e.g. [('a', 1), ('t', 1)] for the word "at"), and
     #      it's important we do this, because a dictionary's keys are NOT SORTED
     #      (so "at" might give us Counter({'t': 1, 'a': 1}), OR Counter({'a': 1, 't': 1}), and we want both
     #       orderings to be equivalent, because they both represent the same letter counts!), then
     # (3) We can turn that (now sorted) list into a string representation (e.g. "[('a', 1), ('t', 1)]", so it
     #      can be used as a _dictionary key_ (otherwise, it's a list, which cannot, because its MUTABLE!)
    if not counter in anagrams:          # Do we already have this counter in our dictionary of letter counts?
        anagrams[ counter ] = []         #  If not, add it, with an empty list as its value (so we can add to it:)
    anagrams[ counter ].append( key )    # Add the word to the list of words associated with this letter count
                                         #  (remember, every word with the same letter count is an anagram!)

## TEST CODE: Let's print out our "anagrams" dictionary, to check that it appears to be as we intend
#print( anagrams )
## It looks okay; we can comment it out now
```

```python
# 4. Add code which iterates over your finished dictionary, and counts the total number of anagrams found. How many
#  are there in the whole of "words.txt"?
#
# You might do this by simply iterating over the dictionary and noticing where a set of letter counts are related
#  to more than one word, or by iterating over it and deleting any entry which only has one word associated with it
#  (then finding out how many keys remain).
# 
# You could even iterate over your list of anagrams, and print out each group, so you can see what your code
#  discovered!
```

```python
# Method 1:
# Iterating over the anagrams dictionary to find which entries have more than one word associated with them
#  (if there is more than one, we have an anagram!)

total_anagrams = 0                       # We'll need an integer, to keep track of how many anagrams we find
for count in anagrams:                   # Iterate over (the keys of) the anagrams array
    if len( anagrams[ count ] ) > 1:     #  Check the length of the list of words associated with this count;
        total_anagrams += 1              #   If it's more than one, it's an anagram (add to our total count)
print( "There are", total_anagrams, "anagrams in total." )  # Print out the total in an easy-to-read fashion


# Method 2:
# Iterating over the anagrams dictionary, and removing all entries with one one associated word
#  This makes sense to me, because if there is only one associate word, there is no anagram!

list_of_counts = list( anagrams.keys() ) # We need to make a COPY of the keys of the dictionary, because if we
                                         #  modify the dictionary using "del", we'll be changing its size whilst
                                         #  we're STILL trying to ITERATE OVER IT; that's a bad idea (it will error)!
for count in list_of_counts:             # Iterate over (our copy of) the dictionary's keys (letter counts)
    if len( anagrams[ count ] ) < 2:     #  If the list of associated words is less than 2, we don't have an anagram
        del anagrams[ count ]            #   ...so we'd better _delete_ that entry from our dictionary!

total_anagrams = len( anagrams )         # Now we've removed all but anagrams from our dictionary, we can simply
                                         #  take its size as the count of total anagrams found
print( "There are", total_anagrams, "anagrams in total." )  # Print out the total again in an easy-to-read fashion

# Note that both methods return the same count, so both are achieving the same outcome


# Note also that we haven't taken account of _how many_ anagrams there are for each letter count!
#  We could do that; rather than taking each GROUP of anagrams as a count of "1 anagram found",
#  we could take the count as the total number of combinations (without repetition) possible given
#  the total number of words in each group of anagrams, taking an anagram to be any pair of words in
#  that group.
#  e.g. if we have 3 words in the list, we could calculate 3! / (2!(3-2)!), which gives us 3 combinations
#  (for instance, "glean" and "angle", "glean" and "angel", "angle" and "angel" = 3 possible combinations).
# You could adapt your code to count the total number of anagrams this way simply by altering this cell's code,
#  for instance by changing the line "total_anagrams = len( anagrams )" above to the following:

from math import factorial as f          # We'll use the factorial() function from Python's math module (and
                                         #  alias it as "f", so we can use "f()" instead of "factorial()" each time
total_anagrams = 0                       # We'll need to start counting from scratch, so reset our counter
for anagram in anagrams:                 # Iterate over each key of our "anagrams" dictionary
    n = len( anagrams[ anagram ] )       #  First, find the number of words discovered for this letter count (n)
    total_anagrams += int( f(n) / ( f(2) * f( n - 2 ) ) )  # Now, calculate n!/(2!(n-2)!), and turn the result into
                                                   #  an integer (we know the result will always be a round number)
    
print( "There are", total_anagrams, "anagrams _combinations_ in total." )  # Print out the total combinations
    
```
