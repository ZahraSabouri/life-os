```python
# EXERCISE 5
#
# Four-of-a-kind Game:
#
# A standard pack of playing cards has four suits: Hearts, Spades, Diamonds and Clubs. In each suit, there are 13
#  ranks: 9 numbered cards (2-10), and 4 picture cards (Ace, Jack, Queen, King). Let's make a game where a player
#  attempts to draw four-of-a-kind (i.e. four cards of the same rank) in a certain number of draws from the deck.
#  For example, given 10 draws, it might be that the player draws three 4s (of any suit) amongst other cards, but
#  not a fourth 4, and so they will not win.
```

```python
# 1. Add the following lines of Python to begin with:
# 
# suits = ( "H", "S", "D", "C" )
# ranks = tuple( range( 1, 14 ) )
# cards = ()
# 
# Let's represent each card as a string. For example, the Ace of Hearts might be "H1", and the Eight of Diamonds
#  might be "D8".
# 
# Now, write a function "add_cards" which takes a single suit string (H, S, D or C) as an argument, and adds all
#  the ranks of that suit to the cards tuple. Remember: "cards" cannot be written to from within a function unless
#  it is considered global, and a tuple cannot be added to, only a whole new tuple created. (We should have used a
#  list here, really; if you find this part of the exercise too difficult, feel free to change "cards" to be a list
#  instead).
```

```python
suits = ( "H", "S", "D", "C" )
ranks = tuple( range( 1, 14 ) )      # This line creates a list of numbers 1-14 (using range()), and stores it in
cards = ()                           #  a tuple. "cards = ()" is equivalent to "cards = tuple()", either is fine.

# add_cards: takes a single character string "suit",
#  and adds all the ranks for that suit to the "cards" tuple
def add_cards( suit ):
    global cards                     # If we don't declare this a global, our function will treat "cards" as local!
                                     #  And, since we haven't declared "cards" locally, we'll get an error.
    for rank in ranks:               # Iterate over all the ranks (we're only reading "ranks"; no need for "global")
        card = suit + str( rank )    #  Create a single card string, from "suit" and a string version of a rank
        cards = cards + ( card, )    #  Add that card string to our cards tuple, by _reassigning it_
                                     #  NOTE: we MUST add the "," after "card", or Python won't take this as a tuple!

## TEST CODE: I want to see whether one call to add_cards() does what I expect it to do:
#add_cards( "H" )
#print( cards )    # I'm expecting to see everything from "H1" to "H13" in this tuple... and I do!
## Now I know it's working, I'll comment out the above code, ready for the next part of the exercise.
```

```python
# Don't forget that if you run this block of code more than once, it will add more and more cards
#  to your "cards" tuple! (Re-run the cell above, where "cards" is set to be empty, to start again.)

for suit in suits:                   # Iterate over all our "suits" strings (H, S, C and D)
    add_cards( suit )                #  For each, call "add_cards" (which adds suit_1 to suit_13 to "cards" for us)

## TEST CODE: Again, I want to see if what I expect to happen has happened;
##  the "cards" tuple should now contain ALL the suits (1-13, for H, S, D and C)
#print( cards )    # It does! Now I know it works, I can comment out this test again.
```

```python
deck = list( cards )                 # If we pass a tuple into list(), we'll get a list-version of it back as a COPY
hand = list()                        # This just creates an empty list (we could have written "hand = []", instead)

## TEST CODE: Again, I'm going to print out both, to make sure I'm getting the two lists I wanted
#print( deck )
#print( hand )
## Looks good! I'll comment out my test once more.
```

```python
# 4. Next, write a function called "take_card", which picks a card at random from the "deck" list, adds it to the
#  "hand" list, and removes it from the "deck" list. Remember, you can import the "random" module, and use
#  "random.choice( <your_list> )" to get a random element from your list.
```

```python
import random

# take_card: has no arguments. It does the following:
#  (1) Picks a single element at random from the "deck" list;
#  (2) Adds that element to the "hand" list; and
#  (3) Removes that element from the "deck" list.
#  This simulates a card moving from the deck to the hand
def take_card():
    card = random.choice( deck )
    hand.append( card )
    deck.remove( card )

## TEST CODE: Does this do what I want? I'll print the deck and hand, call take_card() once, then check them again...
#print( deck )
#print( hand )
#take_card()
#print( deck )
#print( hand )
## Yes, that worked; I now have one card in "hand", and that card has been removed from "deck". Commenting out...
```

```python
# check_win: takes no arguments. It uses the "global" variable "hand" to
#  check whether the player has four string cards of the same rank in
#  their hand (because we're not writing to "hand", we don't need the
#  "global" keyword)
def check_win():
    counts = {}                          # Let's create a dictionary to hold our counts in for each rank
    for rank in ranks:                   # We can iterate over all the rank values (1 to 13)
        counts[ str( rank ) ] = 0        #  and, at each iteration, add a zero count for each (string) rank
    for card in hand:                    # Now we have a dictionary with counts of zero in it, iterate over the hand
        rank = card[ 1 : ]               #  Store a rank; that is every character from the second on (e.g. 13 of H13)
        counts[ rank ] += 1              #  Using tht stored rank as a key into "counts", add one to the total

    return 4 in counts.values()          # We _could_ iterate over the counts array to see if any 4-of-a-kind were
                                         #  seen... or we could get all counts (using .values()), and check whether
                                         #  any value is the number 4; we don't care _which_, _any_ 4 is a win

if check_win():                          # Call our function; did it return True (denoting a win)?
    print( "You won the game!" )         #  If it did, tell the user they won
else:                                    # Otherwise, they must have lost...
    print( "You did not win." )          #  ...so we should inform them of that

    
# I tried re-running my code for questions 3, 5 and 6 several times, with the following values for
#  number_of_draws:
#   10, 15, 20, 25, 30
# I chose those numbers because I thought it would give me a good spread of values, without needing
#  to observe a massive amount of samples.
# I discovered to win most of the time, 25 draws seemed good.
# I tried to refine this, using number_of_card = 26, and that seemed better still.
# So, my final (experimental) answer is that 26 draws are needed in order for the player to win
#  the game most of the time (i.e. more than 50% of all plays).
```
