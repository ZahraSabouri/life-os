```python
# 5. A cinema film from the latter half of the 20th century would be shown at 24 frames (individual pictures) per
#  second. However, a television from that period shows 25 individual pictures per second. In order to enable films
#  to be shown on television without flickering, it was common to transmit televised versions of films at 25 frames
#  per second.
#  (a) How much faster, in percent, would a televised version of film be running, than its cinema counterpart?
#  (b) How long would a televised version of a film last if the original cinema film lasted for 1 hour and 28
#   minutes?
#  (c) How long would an original cinema film have lasted, if its televised version played for 2 hours and 12
#   minutes?
```

```python
# 5(a)
ratio = 25 / 24
increase_percent = ( ratio - 1 ) * 100
print( "A televised film will run", increase_percent, "% faster than the original speed" )
```

```python
# 5(b)
film_time = (1*60) + 28 # In minutes
result = film_time / ratio
print( "The televised film would run for", result, "minutes (the cinema film would run for", film_time, "minutes)" )
```

```python
# 5(c)
tv_time = (2*60) + 12 # In minutes
result = tv_time * ratio
print( "The original film would run for", result, "minutes (the televised film would run for", tv_time, "minutes)" )
```

```python
# (a) should be 0.3, and (b) should be 0.1.
# This happens because of the way decimal precision is handled within the computer,
#  specifically the accuracy with which Python represents floating-point numbers
#  (as many languages do). It is possible to get perfect precision for a specific
#  fraction, but not for _all_ fractions.
# Is this difference significant? It might be; it depends upon how much accuracy
#  your calculation needs. For most purposes, no, there is no issue. Let's
#  work out the actual error:
print( "(a):", ( 0.1 * 3 ) - 0.3 )
print( "(b):", ( 0.3 / 3 ) - 0.1 )
# Both these amounts are very small indeed! For most purposes, the accuracy we
#  see is close enough.
```
