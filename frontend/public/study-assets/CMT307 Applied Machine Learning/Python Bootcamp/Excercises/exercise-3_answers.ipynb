```python
# 3. Given an array of numbers and a number k, write Python code to check whether the sum of any two
#  numbers from the list is equal to k or not. For example, given [1, 5, 11, 5] and k = 16, your program should
#  return "true", since 11 + 5 is 16.
```

```python
numbers = [1, 5, 11, 5]   # An array to contain our numbers. We know 11+5 = 16, so this should return True
k = 16                    # Our variable "k", which we can adjust as we wish (e.g. k=17 should return False)

found = False             # How will we know if we found a combination matching k? Let's keep a boolean flag

for a in numbers:         #  Iterate over all the numbers (here, variable "a" will be equal to 1, then 5, 11 etc.)
    for b in numbers:     #   Iterate AGAIN over all the numbers (so we have a variable "b" equal to 1, 5, 11 etc.)
        if a + b == k:    #    On the first iteration we have a=1, b=1; the second, a=1, b=5, so we can check a+b==k
            found = True  #     If a+b were equal to k, let's set our "yes, it matched!" flag to True (for later)

if found:                 # Did our nested iteration over "numbers" find a combination a+b==k?
    print( "True!" )      #  If so, let's tell the user
else:                     # Otherwise, it didn't find a matching combination
    print( "False." )     #  So, let's tell the user nothing was found
```

```python
# 4. Write a function which accepts a String as input, and returns true if that String is a palindrome, false if
#  not (a palindrome is a word that is spelt the same backwards and forwards).
```

```python
# It's common practice to place functions above the "main" code in a script, so that the function is
#  available (i.e. it's been fully defined) when we need to call it:

# is_palindrome: accepts a single string "s", reverses it into a copy "s2",
#  and then returns True if "s" and "s2" are equal
def is_palindrome( s ):
    s2 = ""                                     # Let's begin with s2 as an empty string
    for i in range( len( s ) - 1, -1, -1 ):     # Let's iterate BACKWARDS over s (range( end, start, step=-1 ) )
        s2 += s[ i ]                            #  At each iteration, we add a character of s to s2 (in reverse)
    
    return s == s2                              # Now we have a reversed s (s2), we can check whether they're equal


# Our "main" code:

s = input( "Word: " )                           # Get a word from the user, and store it in "s"
if is_palindrome( s ):                          # Use our function to discover if "s" is a palindrome
    print( "Palindrome!" )                      #  If it is, report that to the user
else:                                           # Otherwise, it's not a palindrome
    print( "Not a palindrome" )                 #  Let the user know that it wasn't a palindrome
```
