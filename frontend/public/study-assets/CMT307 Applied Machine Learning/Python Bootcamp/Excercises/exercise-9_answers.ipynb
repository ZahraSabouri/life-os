```python
# 1. Given the following list of potential e-mail addresses:
#
#  [ "HumphreysDJ@cardiff.ac.uk", "support@python.org", "support.python@org", "WilliamsPT@programs.com",
#    "AliyevOR@information.gr", "SatoUE@aberdare.ac.uk", "myemail.edu" ]
#
# Use regular expressions to write separate functions which can do the following:
#  (a) Detect whether an e-mail address is "valid" (simply, it must contain two parts separated by an "@" symbol,
#       and at least one suffix following a period ("."),
#       i.e. <username>@<domain>.<tld>
#  (b) Extract the username portion of each e-mail address, and return it;
#  (c) If the username portion has the form <Surname><Initial 1><Initial 2> (e.g. "HumphreysDJ"), extract the
#   surname and both initials, and return them in a list;
#  (d) Detect whether the address ends in an educational suffix (".edu" or ".ac.uk").
#
# Add code which calls your functions on each e-mail address, and prints out the information returned by them
#  (validity, username, surname and intiials is available, and whether the address is an educational one.
```

```python
import re                                                   # We'll need to import Python's Regular Expressions
                                                            #  module before we can access its methods
    
# Let's copy the address list in first, so we can iterate over it and test our code
addresses = [ "HumphreysDJ@cardiff.ac.uk", "support@python.org", "support.python@org", "WilliamsPT@programs.com", \
              "AliyevOR@information.gr", "SatoUE@aberdare.ac.uk", "myemail.edu" ]

# 1(a)
# address_is_valid: takes a string e-mail address, and returns True
#  if it is considered valid, and False if not
def address_is_valid( adrs ):
    return re.match( r"\w+@\w+\.\w+", adrs ) != None        # This regular expression means: (1) match one or more
                                                            #  "word" (alphanumeric/underscore) characters, (2) match
                                                            #  an "@", (3) match one or more "word" characters. (4)
                                                            #  match a ".", (5) match one or more "word" characters.
                                                            # If there is no match, re.match() returns "None"
# 1(b)
# get_username: takes a string e-mail address, and returns the
#  beginning portion of it, before the "@" symbol. If there is
#  no username portion, returns the string "(none)"
def get_username( adrs ):
    regex = r"\w+\.*\w*@"                                   # This means: (1) match one or more word characters, (2)
                                                            #  zero or more "." characters, (3) zero or more word
                                                            #  characters, (4) match an "@" symbol (end of username)
    if re.match( regex, adrs ):                             # Is there a match for this (i.e. does a username exist)?
        return re.findall( regex, adrs )[ 0 ][ : -1 ]       #  If so, find all occurences, and return the _first_ one
                                                            #  (the index [0]), but lose the last character (the "@")
    return "(none)"                                         # If we reach here, we didn't have a match, so return the
                                                            #  string "(none)" instead (this seems appropriate)
    
# 1(c)
# get_surname_and_initials: takes a string e-mail address, and
#  returns a list of strings [ surname, first_initial, second_initial ].
#  If there is no surname and initials present, returns the string "(none)".
def get_surname_and_initials( adrs ):
    regex_surname  = r"[A-Z]+[a-z]+"                        # This means: (1) match one or more capital A-Z letters,
                                                            #  (2) match one or more lower-case a-z characters
    regex_initials = r"[A-Z]{2}"                            # This means: (1) match exactly two capital A-Z letters
    if re.match( regex_surname + regex_initials, adrs ):    # Is there a match for "surname" followed by "initials"?
        surname  = re.findall( regex_surname,  adrs )[ 0 ]  #  If so, fetch the first match for the surname,
        initials = re.findall( regex_initials, adrs )[ 0 ]  #  and fetch the first match for the two initials
        return [ surname, initials[ 0 ], initials[ 1 ] ]    #  Now, return the surname, & both initials (one by one)
    return "(none)"                                         # If we reach here, we didn't have a match, so return the
                                                            #  string "(none)" instead

# 1(d)
# has_educational_suffix: takes a string e-mail address, and returns
#  True if it has the suffix ".ac.uk" or ".edu", and False otherwise
def has_educational_suffix( adrs ):
    return re.search( r"(.edu)|(.ac.uk)", adrs ) != None    # This regular expression means: (1) match the whole term
                                                            #  ".edu", (2) OR (the "|" symbol), (3) match the whole
                                                            #  term ".ac.uk". If there is no match (for either term)
                                                            #  re.match() returns "None" (so we return False)
            
# Let's add some code to loop over all the e-mail addresses, and print out
#  the results of the above functions for each:
for adrs in addresses:
    # We'll try to format our print-out nicely; we can specify a column width for each string we print using
    #  the "%s[num_characters]" form, for instance here we request 5 characters in width for our "Valid?:"
    #  strings (the output of address_is_valid() ), and 14 characters in width for our "Username:" strings
    #  (the output of get_username() ). We're also using the backslash (\) so we can continue our code on
    #  the next line, to make it more readable, since there are a lot of function calls and brackets!
    print( "Valid?: %5s;    Username: %14s;    Surname/Initials: %23s;    Educational?: %5s" % \
           ( address_is_valid( adrs ), \
             get_username( adrs ), \
             get_surname_and_initials( adrs ), \
             has_educational_suffix( adrs ) \
         ) )
```

```python
# Download the archive "articles.zip" from Learning Central (from the "CMT307/Python Bootcamp/Exercises"
#  folder), unzip it on your computer, and upload all the text articles files ("article1.txt" to "article5.txt") to
#  your Google Colab (temporary) file area.
#
# 2. Given these articles, write code which does the following for each article:
#  (a) Loads in the article's text;
#  (b) Uses a regular expression to extract all the words which begin with capital letters, after a whitespace (e.g.
#   " Cardiff" should match, but not "Travel");
#  (c) Uses a regular expression to extract all the numbers which follow a whitespace, and all monetary values /
#   prices which follow a whitespace (e.g. "10,000", "$2.54", "£1.43", "2024" etc.);
#  (d) Prints out the list of all matches for each article (perhaps with a count of the number of occurrences of
#   each string found; you might try the Counter's method ".most_common()" and slicing to fetch only the top 10
#   word counts, or the top 20).
#
# From looking at the returned strings, can you tell:
#  (f) What the general topic of each article is?
#  (g) What subject _all_ the articles come from?
#
# Is it possible to use a technique such as this to understand the contents of articles more quickly than reading
#  them individually?
```

```python
from collections import Counter                      # Let's use a Counter here again, as most_common() will be
                                                     #  useful in counting up the number of occurrences of each word

# 2(a): Load the text of all articles
#  First of all, we can build a list of all our articles' filenames programmatically:
fns = []                                             # Create an empty list
for i in range( 1, 6 ):                              # We have the names "article1.txt" to "article5.txt", so we can
                                                     #  iterate over the numbers "1" to "5" to build these names
    fns.append( "article" + str( i ) + ".txt" )      #  Append "article", plus the number, plus ".txt" to our list
#  Now we have a list of all our filenames, let's iterate over them
#   to load _and_ process each article in turn:
for fn in fns:                                       # Use a for loop to fetch each filename in turn (as "fn")
    print( fn )                                      #  Print out the filename, so we can see what's being loaded
    f = open( fn, "r" )                              #  Open the file in "read" mode ("r"), as handle "f"
    d = f.read()                                     #   and read in ALL text at once (into the variable "d")
    f.close()                                        #  Close the file now that we've finished reading it
#    print( d )                                       #  (We could now print what we've loaded, for testing purposes)

# 2(b): Extract all the words which begin with capital letters, after a whitespace
    found = re.findall( r"\s[A-Z]\w*", d )           #  This means: (1) match exactly one whitespace, (2) match zero
                                                     #   or more occurrences of a "word" character. findall() returns
                                                     #   all matches to the pattern (strings) in a tuple
#    print( found )                                  #  (We could now print what we found, for testing purposes)

# 2(d), for part (b): print out the top 20 list of all matches for this article
    counts = Counter( found ).most_common()[ : 20 ]  #  Place all found words into a Counter, fetch the greatest
                                                     #   counts (most_common()), and keep just the first 20 (slicing)
    print( counts )                                  #  Print out these top 20 (word) matches

# 2(c): Extract all the numbers which follow a whitespace, including all monetary values
    found = re.findall( r"\s[£\$]*\d[\.,]*\d*", d )  #  This means: (1) match exactly one whitespace, (2) match zero
                                                     #   or more "£" or "$" characters, (3) match a numeric digit,
                                                     #   (4) match zero or more "." or "," characters, (5) match zero
                                                     #   or more numeric digits
#    print( found )                                  #  (We could now print what we found, for testing purposes)

# 2(d), for part (c): print out the top 20 list of all matches for this article
    counts = Counter( found ).most_common()[ : 20 ]  #  Again, place all found strings into a Counter, fetch those
                                                     #   with the most occurrences, and keep only the first 20
    print( counts )                                  #  Print out these top 20 (numeric) matches
    #
    
    print( "\n" )                                    #  Let's print a blank line, to separate our output nicely
```
