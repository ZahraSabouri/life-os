```python
# EXERCISE 4
#
# Let's use Python to construct a temporary address book, or contact list, which (for each entry) can contain
#  a person's name, mobile number, house name / number, and postcode, all as strings.
```

```python
# 1. Write Python code which creates an empty dictionary called "contacts". Create a function which takes a name,
#  number, house name and postcode, and adds them to the "contacts" dictionary, using the name as the key (which
#  is the best way to store the rest of the details, as the key's value?).
```

```python
# Create a "global" dictionary to hold our contacts, indexed by name
contacts = {}

# add_contact: takes the arguments name, number, house and postcode,
#  and adds them to an entry with the name as the key,
#  and a dictionary of the remaining data as a value (with keys
#  "number", "house" and "postcode")
def add_contact( name, number, house, postcode ):
    contacts[ name ] = { "number" : number, "house" : house, "postcode" : postcode }
```

```python
# 2. Write a while loop which will run until a user enters the word "quit", and within that loop fetch some input
#  from the user. If user hasn't typed "quit", fetch a new name, number, house name and postcode from them, and
#  use those details to add to the dictionary. Make you code print out all the names in "contacts" after each
#  addition. You should now have a program which will keep asking the user for new contacts to enter in the address
#  book, and which will quit when the user wishes.
```

```python
# Let's use an infinite loop (while True), and stop it by using the keyword
#  "break" when required (i.e. when the user has typed "quit"):
while True:
    cmd = input( "(quit) >: " )                   # Let's show the user the "quit" keyword option, to be helpful
    if cmd == "quit":                             # If they typed "quit"...
        break                                     #  then break out of the loop (which also ends the program)
    name     = input( "Name: " )                  # Otherwise, request a name, number, house name and postcode
    number   = input( "Number: " )
    house    = input( "House name: " )
    postcode = input( "Postcode: " )
    add_contact( name, number, house, postcode )  # Use our add_contact() function to add them to the contacts dict
```

```python
# 3. Change your code so that the user may type "search" instead of "quit". If they do, have your program ask them
#  to enter a name to search for. If the name they enter exists in the dictionary, print out that contact's
#  details. If not, print a message to let the user know the person has not been found.
```

```python
# Let's put our searching code in a function, as it'll be tidier, and easier to use within our (already quite
#  full) loop

# search_for: takes a string name as an argument,
#  and either prints out the contact details of the entry with that name
#  (in a nicely-formatted output), or prints "Name not found."
def search_for( name ):
    if not name in contacts:                     # Does the name exist as a key in the contacts dictionary?
        print( "Name not found." )               #  If not, we'll print "Name not found."...
        return                                   #  ...and then immediately quit the function
    contact = contacts[ name ]                   # Otherwise, fetch the contact's details, and print them out
    print( "Name: %s, Number: %s, House name: %s, Postcode: %s" % ( name,               contact[ "number" ], \
                                                                    contact[ "house" ], contact[ "postcode" ] ) )

while True:
    cmd = input( "(quit/search) >: " )           # We've added "search" as a keyword the user can enter, too
    if cmd == "search":                          # Did the user type "search"?
        name     = input( "Name? " )             #  If they did, fetch a name to search for...
        search_for( name )                       #  ...and run our "search_for" function, giving it that name
    elif cmd == "quit":                          # OTHERWISE, did the user type "quit"?
        break                                    #  If so, break out of the loop (and the program will end)
    else:                                        # OTHERWISE, request new contact details, and add them to the
        name     = input( "Name: " )             #  contacts dictionary
        number   = input( "Number: " )
        house    = input( "House name: " )
        postcode = input( "Postcode: " )
        add_contact( name, number, house, postcode )  
```

```python
# 4. Try adapting your code so that the user must enter "add" if they wish to add an entry, and if they do not
#  enter one of the words "quit", "search" or "add" the program will keep looping.
```

```python
while True:
    cmd = input( "(quit/search/add) >: " )      # We've added "add" as a keyword to prompt the user this time
    if cmd == "search":
        name     = input( "Name? " )
        search_for( name )
    elif cmd == "quit":
        break
    elif cmd == "add":                          # We've changed "else" for a test to see if the user typed "add"
        name     = input( "Name: " )            #  If they did, fetch and store the details just as before
        number   = input( "Number: " )
        house    = input( "House name: " )
        postcode = input( "Postcode: " )
        add_contact( name, number, house, postcode )  
```

```python
# Let's move our "print out a contact" code from the "search_for" function into a new function, "show_contact",
#  so that we can re-use it when we search for a contact by number, also

# show_contact: takes a string name as an argument, and prints out details of the contact with that name
#  Note that this function assumes that the contact DOES exist (we're not checking whether they do or not!)
def show_contact( name ):
    contact = contacts[ name ]
    print( "Name: %s, Number: %s, House name: %s, Postcode: %s" % ( name,               contact[ "number" ], \
                                                                    contact[ "house" ], contact[ "postcode" ] ) )

# search_for: takes a string name as an argument,
#  and either prints out the contact details of the entry with that name
#  (in a nicely-formatted output), or prints "Name not found."
def search_for( name ):
    if not name in contacts:                    # Does the contact NOT exist in our dictionary?
        print( "Name not found." )              #  If not, print out a message to show this is the case,
        return                                  #  and then immediately return from (end) the function
    show_contact( name )                        # Otherwise, use our "show_contact" function to print their details

# search_by_number: takes a string number as an argument,
#  and either prints out the contact details of the entry with that name
#  (in a nicely-formatted output), or prints "Number not found."
def search_by_number( number ):
    for name in contacts.keys():                # Iterate over _all_ the contacts in our dictionary (SLOW!)
        if contacts[ name ][ "number" ] == number:  # Is this particular contact's number a match for that sought?
            show_contact( name )                #  If so, use "show_contact" to print their details...
            return                              #  ...and then immediately quit the function (and our for loop)
    print( "Number not found." )                # Otherwise, if this code runs, we didn't find the number
                                            #  (because "return" would have quit our function beforehand, if we did)
while True:
    cmd = input( "(quit/search/numsearch/add) >: " )  # Let's add our new "numsearch" keyword to the prompt list
    if cmd == "search":
        name     = input( "Name? " )
        search_for( name )
    elif cmd == "numsearch":                    # Did the user type "numsearch"?
        number   = input( "Number? " )          #  If they did, fetch a number to search for from them...
        search_by_number( number )              #  ...and then use our "search_by_number" function to find it
    elif cmd == "quit":
        break
    elif cmd == "add":
        name     = input( "Name: " )
        number   = input( "Number: " )
        house    = input( "House name: " )
        postcode = input( "Postcode: " )
        add_contact( name, number, house, postcode )
        
# Using a for loop in "search_by_number" to check the number of _each_ contact will be SLOWER
#  than our search for a name using the "in" keyword (because "in" is likely optimised within the
#  Python interpreter, and we're having to (a) get a new key each time, then (b) retrieve the
#  number associated with that key, as well as using an "if" statement to see whether the number
#  we retrieved matches the one we're seeking).
#
# Yes, we could do something about this! We could keep another dictionary whose keys are "number"
#  values, and whose values are "name" keys from our main "contacts" dictionary. That way, if
#  we want to search by number, we can just use "in" as we did for "contacts" (name search),
#  and then use the "name" we get from our new dictionary to index into our "contacts" one
#  (so we could then access all their details, as normal). However, this would take up MORE
#  MEMORY, because now we've stored all the names and numbers TWICE! But, if we want a fast
#  search, it might be worth it.
#
# NOTE: In simple, small examples such as this, such things do not matter much; the size of
#  our data will be very small, and the delay of iterating over a dictionary instead of using
#  "in" to check for membership of a key is extremely minor. Nonetheless, we should think about
#  such things, because the moment we start dealing with big data, the delay (or the used space,
#  if we choose to maintain a "reverse lookup" number dictionary as well, as discussed above)
#  is likely to become significant.
```
