```python
import pandas as pd                       # We must import pandas before we can use it (using the alias "pd", here)
pok = pd.read_csv( "pokemon.csv" )        # Read CSV data from our file into a DataFrame assigned to the label "pok"

# 1: We can use the ".columns" property to retrieve a list of column names
print( pok.columns )
#     (If we _really_ wanted to make it a Python list, we could do as follows:)
#print( list( pok.columns ) )
```

```python
# 2. Using the "Type 1" column, create a list of unique names in that column. Iterate over those names, and
#  calculate the proportion of all rows which that "Type 1" represents. Which is the largest type in the DataFrame,
#  and what percentage of all entries does it represent?
```

```python
import numpy as np                        # Let's import NumPy, so we can use its methods on our DataFrame

# Part 1: Creating a list of unique names in the "Type 1" column:
type1s = list( set( pok[ "Type 1" ] ) )   # We can get all "Type 1" names, place them in a set (which stores only
                                          #  unique values), then convert it to a list
print( "Type 1s: ", type1s, "\n" )        # Let's print out our list, followed by newline

# Part 2: Iterate over the names, and calculate the proportion of all rows which each represents
type1n = []                                           # Let's create an empty list to hold our list of proportions
for type1 in type1s:                                  # Iterate over each Type 1 name
#  We can do this step by step...:
#     num_of_type = np.sum( pok[ "Type 1" ] == type1 )  #  Sum the number of times the name appears in the Type 1 col.
#     proportion  = num_of_type / pok.shape[ 0 ]        #  Divide that by the total number of rows to get a proportion
#     percentage  = proportion * 100                    #  Multiply the proportion by 100 to get a percentage
#     type1n.append( percentage )                       #  Append that percentage to our list (to match our names list)
#  ...or, we could do it in one compound statement:
    type1n.append( ( np.sum( pok[ "Type 1" ] == type1 ) / pok.shape[ 0 ] ) * 100 )

#  Let's print out our findings, so we can see what percentages we have for each type:
for i in range( len( type1n ) ):                                         # Iterate over all indices in our names list
  print( "Type 1: %s, Percentage: %.1f" % ( type1s[ i ], type1n[ i ] ) ) #  Print a name and a percentage for each

# Part 3: Finding the largest type in the DataFrame (i.e. the one with the most occurrences):
index = np.argmax( type1n )                # First, use np.argmax() to get the index of the greatest percentage
#
print( "\nLargest group found:" )                 # Let's print a newline, and a title, so we know what follows
name_of_largest = type1s[ index ]                 # We can fetch the name from the names list; it has the same order 
print( "\tName:       %s" % name_of_largest )     #  as our percentages list, and the same number of elements
pcnt_of_largest = type1n[ index ]                 # We can of course fetch the greatest percentage using our index
print( "\tPercentage: %.1f%%" % pcnt_of_largest ) # Let's print it out to 1 d.p. and with a suffixed "%" character
```

```python
# 3. Set the DataFrame's index to the column "Type 1", and store this in a new variable ("pok_type1" perhaps?).
#  Selecting only the "Grass" type 1s, calculate the sum of both "Attack" and "Speed" columns for each of these
#  Pokemon. Which is the largest total, and what is the name of the Pokemon to which this belongs?
```

```python
# Part 1: Setting the DataFrame's index to the column "Type 1":
pok_type1 = pok.set_index( "Type 1" )                     # We can use .set_index() to choose a column to index from

# Part 2: Selecting only entries from the "Type 1" column which match the string "Grass",
#          and keeping only the "Name" column from each row:
names = list( pok_type1.loc[ "Grass", "Name" ] )          # Select all the "Grass" rows, then just the "Name" column,
print( names, "\n" )                                      #  using .loc() (we can then turn the result into a list,
                                                          #  and then print it out, with a following newline)

# Part 3: Selecting only the "Attack" and "Speed" columns from each "Grass" entry, and
#          calculating their sum (for each row):
# We can do this in individual steps...:
# att_spd = pok_type1.loc[ "Grass", [ "Attack", "Speed" ] ] # Fetch only the "Attack" and "Speed" columns for each
#                                                           #  "Grass" Type 1 entry
# as_sums = np.sum( att_spd, axis = 1 )                     # Sum up each set of values, row-wise (i.e. one 
#                                                           #  attack+speed sum for each row we have)
# totals = list( as_sums )                                  # Convert our results into a list which we can index
#                                                           #  into in just a moment, and which matches our names list
# ...or we can make one single compound statement:
totals = list( np.sum( pok_type1.loc[ "Grass", [ "Attack", "Speed" ] ], axis = 1 ) )
print( totals, "\n" )                                       # Print out our list of totals, so we can see our results

# Part 4: Finding the largest total, and printing out the name of the associated Pokemon:
index = np.argmax( totals )                               # Fetch the index of the greatest total we discovered
print( "Greatest combination: ", names[ index ], " = ", totals[ index ] )  # Use that index to fetch both the name
                                                                           #  and the total for that row
```

```python
# All we need to do is fetch all rows from the original DataFrame ("pok"),
#  where the string in the "Name" column exactly matches the name string we found above:
print( "From original table:" )
print( pok[ pok[ "Name" ] == names[ index ] ] )           # Index into the original table using our name string

# Yes, the details _do_ match: we have Attack=110 and Speed=145, and 110+145 = 255 (the value calculated above)
```
